```python
from imblearn.over_sampling import RandomOverSampler
```

```python
X = tecal.drop('cancel', axis=1)  # todas as variáveis explicativas
y = tecal['cancel']               # variável resposta
```

```python
tecal_balanceamento = RandomOverSampler(sampling_strategy='auto', random_state=42)
X_resampled, y_resampled = tecal_balanceamento.fit_resample(X, y) # aplicando o balanceamento
```

```python
tecal_balanceado = pd.concat([pd.DataFrame(X_resampled), pd.DataFrame(y_resampled, columns=['cancel'])], axis=1) # juntar novamente em um dataframe balanceado
```

```python
print(tecal_balanceado['cancel'].value_counts())
print(len(tecal_balanceado)) # verificando o novo balanceamento
```

```python
tecal_balanceado = tecal_balanceado.sample(n=3000, random_state=42) # ajustando para ter aproximadamente 3000 obeseravções
```

```python
print(tecal_balanceado['cancel'].value_counts())
print(len(tecal_balanceado))
```
